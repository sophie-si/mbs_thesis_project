```python
#Combine fdr bpm significant tables
#Count how many times BPMs have significant fdr values
import pandas as pd
```

```python
fdrbpm = pd.read_excel("fdrBPM_significant.xlsx")
```

```python
#Go through each column which is the run. 
#Check the fdr value for that
#Add that to the significant runs counter
#once you are done counting, create a df and add that to it
#repeat
```

```python
col_names = list(fdrbpm.columns)
#Get the columns that contain the fdr values
fdr_inds = [i for i, col_name in enumerate(col_names) if 'CRC' in col_name]
fdr_all = pd.DataFrame(columns = ['significant_runs_0.25', 'significant_runs_0.2', 'significant_runs_0.15', 'significant_runs_0.1', 'significant_runs_0.05'])
for i in range(fdr_inds[0],(fdr_inds[-1]+1)):
    fdr_dict = {'significant_runs_0.25':0, 'significant_runs_0.2':0,'significant_runs_0.15':0, 'significant_runs_0.1':0, 'significant_runs_0.05':0}
    for j in range(len(fdrbpm)):
        currfdr = fdrbpm.iloc[j,i]
        if currfdr < 0.05:
            fdr_dict['significant_runs_0.05']+=1 
            fdr_dict['significant_runs_0.1'] +=1 
            fdr_dict['significant_runs_0.15'] +=1 
            fdr_dict['significant_runs_0.2'] += 1
            fdr_dict['significant_runs_0.25']+=1
        elif currfdr < 0.1:
            fdr_dict['significant_runs_0.1'] +=1 
            fdr_dict['significant_runs_0.15'] +=1
            fdr_dict['significant_runs_0.2'] += 1
            fdr_dict['significant_runs_0.25']+=1
        elif currfdr < 0.15:
            fdr_dict['significant_runs_0.15'] +=1 
            fdr_dict['significant_runs_0.2'] += 1
            fdr_dict['significant_runs_0.25']+=1
        elif currfdr < 0.2:
            fdr_dict['significant_runs_0.2'] += 1
            fdr_dict['significant_runs_0.25']+=1
        elif currfdr < 0.25:
            fdr_dict['significant_runs_0.25']+=1
    #add to dataframe
    print(fdr_dict)
    fdr_all.loc[len(fdr_all)] = fdr_dict
```

```python
rownames = fdrbpm.columns[fdr_inds[0]:fdr_inds[-1]+1]
fdr_all.index=rownames
```

```python
fdr_all.to_excel("bpm_fdr_counts.xlsx")
```
