```python
import pandas as pd
#load dataframes in
bpm = pd.read_excel('combined_bpm_fdr_pv.xlsx')
wpm = pd.read_excel('combined_WPM_fdr_pv.xlsx')
```

```python
#get the min_fdrs and stitch those columns for significant runs
col_names = list(bpm.columns)
true_inds = []
min_fdr = []
#use the threshold of you choice to determine which bpms would be significant from the results
#for this, I used bpms that met a p-value threshold of 0.05 for at least 5 out of ten runs 
num_sig_bpms = len(bpm['significant_runs_0.05'][bpm['significant_runs_0.05'] >= 5])
#find which columns have fdr in its name. The fdr columns should be next to each other.
fdr_inds = [i for i, col_name in enumerate(col_names) if 'fdr' in col_name]

#find the min_fdr and add it to the list
for i in range (0,num_sig_bpms):
    min_fdr.append(min(bpm[col_names[fdr_inds[0]:(fdr_inds[-1]+1)]].iloc[i]))
    true_ind = bpm[col_names[fdr_inds[0]:(fdr_inds[-1]+1)]].iloc[i]==min(bpm[col_names[fdr_inds[0]:(fdr_inds[-1]+1)]].iloc[i])
    true_ind1 =true_ind[true_ind==True].index
    true_inds.append(true_ind1)
```

```python
new_bpm = bpm.loc[range(num_sig_bpms), ['path1', 'path2', 'disease_association', 'significant_runs_0.05']]
new_bpm['min_fdr'] = min_fdr
new_bpm['inds'] = true_inds
```

```python
#write the df to output
new_bpm.to_csv("top_bpms.csv", index=False)
```

```python
#repeat for wpms
#get the min_fdrs and stitch those columns for significant runs
col_names = list(wpm.columns)
true_inds = []
min_fdr = []
fdr_inds = [i for i, col_name in enumerate(col_names) if 'fdr' in col_name]
for i in range (len(wpm)):
    min_fdr.append(min(wpm[col_names[fdr_inds[0]:(fdr_inds[-1]+1)]].iloc[i]))
    true_ind = wpm[col_names[fdr_inds[0]:(fdr_inds[-1]+1)]].iloc[i]==min(wpm[col_names[fdr_inds[0]:(fdr_inds[-1]+1)]].iloc[i])
    true_ind1 =true_ind[true_ind==True].index
    true_inds.append(true_ind1)
```

```python
new_wpm = wpm[['pathway', 'disease_association', 'significant_runs_0.05']]
new_wpm['min_fdr'] = min_fdr
new_wpm['inds'] = true_inds
```

```python
#write to output
new_wpm.to_csv("top_wpms.csv", index=False)
```
