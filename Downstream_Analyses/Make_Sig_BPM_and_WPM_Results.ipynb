```python
#Combine fdr bpm significant and bpm pv significant tables
#show BPMs that have fdr less than or equal to 0.25 and p-value less than or equal to 0.05.
import pandas as pd
```

```python
#load dataframes in
fdrbpm = pd.read_excel("fdrBPM_significant.xlsx")
bpmpv = pd.read_excel("bpm_pv_significant.xlsx")
```

```python
#use the fdr df and get the value from the first column to know what the index is and look that up on 
#pv table. If it's less than 0.05, it will be added to the new table. just look at the sig run column 
# Then make a new df that has the index, disease association, path1, path2, CRC_g1-10, sigrun .25

#Make a df that includes the fdr values for the bpms for all 10 runs. Also extract the column for how many times it met a sig threshold of 0.25.
all_df = fdrbpm.copy()
column_names = all_df.columns.to_list()
CRC_nameslist = []
for i in column_names:
    if "CRC" in i:
        all_df.rename(columns={i: i+'_fdr'}, inplace=True)
        CRC_nameslist.append(i)
#Remove columns you don't need
all_df.drop(columns=['significant_runs_0.2', 'significant_runs_0.15', 'significant_runs_0.1', 'significant_runs_0.05'], inplace=True)
CRC_nameslist.append("significant_runs_0.05")
```

```python
#This will be stitched together after it's grown. Use this to extract the p-values for the BPMs for the 10 runs and
#number of times it met a sig threshold of 0.05
addlaterdf = pd.DataFrame(columns=CRC_nameslist)

#Get the indices that don't have what you are looking for
wrongcolsinds = set()
for j in CRC_nameslist:
    for i in range(len(bpmpv.columns)):
        if j == bpmpv.columns[i]:
            wrongcolsinds.add(i)
allcolinds = set()
for i in range(len(bpmpv.columns)):
    allcolinds.add(i)
corrcolinds = allcolinds.symmetric_difference(wrongcolsinds)
corrcolinds = list(corrcolinds)

#Get the values that you want from the bpmpv df
for i in range(len(fdrbpm)):
    currind = fdrbpm.iloc[i,0]
    currow = fdrbpm.iloc[i]
    currcol = bpmpv.iloc[:,0].tolist()
    currpv = bpmpv.iloc[[currcol.index(currind)]]
    #remove columns you won't be adding to the new dataframe
    currpv = currpv.drop(columns = currpv.columns[corrcolinds])
    if currpv["significant_runs_0.05"].iloc[0] >= 1:
        #Add to the dataframe and grow it
        currpv = currpv.iloc[0].tolist()
        addlaterdf.loc[i] = currpv  
```

```python
#Rename them
#Add the dataframe you just made to the all_df, so it contains all the data you need
for i in column_names:
    if "CRC" in i:
        addlaterdf.rename(columns={i: i+'_pv'}, inplace=True)
df_combined = pd.concat([all_df, addlaterdf], axis=1)
```

```python
#write to output
df_combined.to_excel("combined_bpm_fdr_pv.xlsx", index=False) 
```
