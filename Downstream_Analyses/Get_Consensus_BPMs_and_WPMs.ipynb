```python
#filter out p-values from the combined BPM and WPM files. This information will be used to make network plots.
import pandas as pd
```

```python
bpmin = pd.read_excel("combined_bpm_fdr_pv.xlsx")
bpm_all = pd.DataFrame(columns = ["disease association", "path1", "path2"] )
col_names = bpmin.columns
col_names_list = col_names.to_list()
counter = 0
#Filter out p-values that don't reach 0.05 threshold for less than 5 out of 10 runs
for i in range (len(bpmin)):
    curr_pv = bpmin.iloc[i, col_names_list.index("significant_runs_0.05")]
    if curr_pv >= 5:
        #select the columns that contain disease association and the pathway names
        bpm_all.loc[counter] = bpmin.iloc[i, 1:4].values
        counter += 1
print('Number of Significant BPMs:', counter)
```

```python
#Export it
bpm_all.to_excel("BPM_pval_filter.xlsx")
```

```python
#Repeat for WPMs
wpmin = pd.read_excel('combined_WPM_fdr_pv.xlsx')
wpm_all = pd.DataFrame(columns= ["disease association", "pathway"])
col_names = wpmin.columns
col_names_list = col_names.to_list()
counter = 0
for i in range(len(wpmin)):
    curr_pv = wpmin.iloc[i,col_names_list.index("significant_runs_0.05")]
    if curr_pv >= 5:
        wpm_all.loc[counter] = wpmin.iloc[i,1:3].values
        counter += 1
print("Number of significant WPMs:", counter)
```

```python
#Export it
wpm_all.to_excel("WPM_pval_filter.xlsx")
```
