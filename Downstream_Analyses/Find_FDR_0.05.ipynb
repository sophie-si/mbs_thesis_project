```python
#Combine fdr bpm significant tables
#Count how many times BPMs have significant fdr values
import pandas as pd
```

```python
fdrbpm = pd.read_excel("fdrBPM_significant.xlsx")
fdr_all = pd.DataFrame(columns = ['disease association', 'path1', 'path2'])
```

```python
col_names = fdrbpm.columns
col_names_list = col_names.to_list()

#select bpms that meet a threshold of 0.05 for FDR at least once
for i in range(len(fdrbpm)):
    currfdr = fdrbpm.iloc[i,col_names_list.index("significant_runs_0.05")]
    if currfdr >= 1:
        #1:4 are the indices for the disease association and the bpm names
        fdr_all.loc[i] = fdrbpm.iloc[i,1:4].values
```

```python
#write to output
fdr_all.to_excel("bpm_fdr_0.05.xlsx")
```

```python
#repeat for wpms
fdrwpm = pd.read_excel("fdrWPM_significant.xlsx")
fdr_all2 = pd.DataFrame(columns = ['disease association', 'pathway'])
col_names = fdrwpm.columns
col_names_list = col_names.to_list()
fdr_counter = 0
for i in range(3,13):
    for j in range(len(fdrwpm)):
        currfdr = fdrwpm.iloc[j,i]
        if currfdr < 0.05:
            fdr_all2.loc[fdr_counter] = fdrwpm.iloc[j,1:3].values
            fdr_counter += 1
fdr_all2 = fdr_all2.drop_duplicates(subset='pathway', keep= 'first')
fdr_all2.to_excel("wpm_fdr_0.05.xlsx")
```
